## 🎯 What is an Outlier?
An outlier is an observation that lies an abnormal distance from other values.
- **Example:** Average salary is $60,000, but a CEO earns $15,000,000.
- If you train a Linear Regression model with this CEO row, the slope gets dragged massively upward, ruining predictions for the 99% normal employees!

### The IQR (Interquartile Range) Method:
$$IQR = Q_3 - Q_1$$
- **Lower Threshold:** $Q_1 - 1.5 \times IQR$
- **Upper Threshold:** $Q_3 + 1.5 \times IQR$

```python
import numpy as np
import pandas as pd

np.random.seed(42)
# 100 normal values around 50, plus 3 extreme anomalies (outliers)
data = np.concatenate([np.random.normal(50, 5, 100), [150, 175, -20]])
df = pd.DataFrame({"metric": data})

q1 = df["metric"].quantile(0.25)
q3 = df["metric"].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print(f"25th Percentile (Q1) : {q1:.2f}")
print(f"75th Percentile (Q3) : {q3:.2f}")
print(f"IQR                  : {iqr:.2f}")
print(f"Valid Range          : [{lower_limit:.2f}, {upper_limit:.2f}]")

# Detect outliers
outliers = df[(df["metric"] < lower_limit) | (df["metric"] > upper_limit)]
print(f"\nDetected {len(outliers)} Outlier Values:\n", outliers["metric"].values)

# Remediation: Winsorization (Clipping)
# Instead of deleting rows, cap them at the thresholds!
df["metric_clipped"] = df["metric"].clip(lower=lower_limit, upper=upper_limit)
print("\nBefore Clipping - Max Value:", df["metric"].max())
print("After Clipping  - Max Value:", df["metric_clipped"].max())
```
