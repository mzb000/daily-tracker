## 🎯 Series vs DataFrame
- **Series:** A 1-dimensional labeled array (like a single column of an Excel spreadsheet with an index).
- **DataFrame:** A 2-dimensional labeled data structure with columns of potentially different types (like a whole Excel table or SQL table).

```python
import pandas as pd

# Creating a DataFrame from a dictionary
dataset = {
    "employee_id": [101, 102, 103, 104],
    "name": ["Aaliyah", "Bilal", "Chaudhry", "Danyal"],
    "department": ["Engineering", "Data Science", "Data Science", "Product"],
    "salary": [85000, 92000, 97000, 78000],
    "remote": [True, True, False, True]
}

df = pd.DataFrame(dataset)
print("DataFrame Head:")
print(df.head())
print("\nSummary Info:")
df.info()
```

### Step 2: Selecting Data: `loc` vs `iloc`
The #1 source of beginner confusion:
- **`iloc` (Integer-location):** Selects by numerical position (0, 1, 2...), exactly like Python lists.
- **`loc` (Label-location):** Selects by row/column index names.

```python
# Select the first 2 rows and columns 1 to 3 by position
print("Using iloc (Row 0-1, Col 1-2):")
print(df.iloc[0:2, 1:3])

# Select employees with salary > 80,000 using loc
print("\nUsing loc with condition (Salary > 80k):")
print(df.loc[df["salary"] > 80000, ["name", "department", "salary"]])
```
