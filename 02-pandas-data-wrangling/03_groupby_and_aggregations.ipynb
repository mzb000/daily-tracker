## 🎯 The "Split-Apply-Combine" Paradigm
1. **Split:** Divide the table into groups based on key columns (e.g., Department, City).
2. **Apply:** Compute metrics independently for each group (Mean, Sum, Count, Min, Max).
3. **Combine:** Merge results back into a clean summary table.

```python
import pandas as pd

sales_data = pd.DataFrame({
    "region": ["North", "North", "South", "South", "East", "East", "North"],
    "product": ["Laptops", "Phones", "Laptops", "Phones", "Laptops", "Phones", "Laptops"],
    "revenue": [12000, 8500, 14000, 7200, 9500, 6100, 11000],
    "units": [10, 15, 12, 14, 8, 11, 9]
})

# Multi-metric GroupBy
summary = sales_data.groupby("region").agg({
    "revenue": ["sum", "mean"],
    "units": "sum"
})
print("Region Sales Summary:")
print(summary)

# Pivot Table: Cross-tabulate Region vs Product
pivot = pd.pivot_table(
    sales_data,
    values="revenue",
    index="region",
    columns="product",
    aggfunc="sum",
    fill_value=0
)
print("\nPivot Table (Revenue by Region and Product):")
print(pivot)
```
