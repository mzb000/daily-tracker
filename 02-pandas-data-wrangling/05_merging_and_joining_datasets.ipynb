## 🎯 Real-World Relational Data
In real life, data is split across multiple tables:
1. `users` table (User ID, Age, City).
2. `transactions` table (Transaction ID, User ID, Amount).

To train an ML model predicting customer churn, we must **join** these tables on their common key (`user_id`).

```python
import pandas as pd

users = pd.DataFrame({
    "user_id": [1, 2, 3, 4],
    "name": ["Zara", "Ahmed", "Fatima", "Hamza"],
    "city": ["Lahore", "Karachi", "Islamabad", "Peshawar"]
})

transactions = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004],
    "user_id": [1, 2, 2, 5],  # Notice user 5 does not exist in users table
    "amount": [150.0, 320.0, 95.0, 450.0]
})

# 1. Left Merge: Keep all users, attach order details where available
left_join = pd.merge(users, transactions, on="user_id", how="left")
print("Left Join (All users kept):")
print(left_join)

# 2. Inner Merge: Keep only users who placed orders
inner_join = pd.merge(users, transactions, on="user_id", how="inner")
print("\nInner Join (Only matching rows):")
print(inner_join)
```
