## 🎯 Why Do Datasets Have Missing Values?
- Sensor transmission errors or timeouts.
- Optional user profile fields (e.g., users skipping phone number).
- Data corruption during database migrations.

### The 3 Options:
1. **Drop Rows (`dropna`):** Dangerous if data is scarce (you lose valuable training examples).
2. **Mean / Median Imputation:**
   - **Mean**: Good for symmetric, normally distributed data.
   - **Median**: **Best for skewed data with outliers** (e.g., Salary, House Prices).
3. **Mode Imputation:** For categorical features (most frequent category).

```python
import numpy as np
import pandas as pd

raw_records = {
    "patient_id": [1, 2, 3, 4, 5, 6],
    "age": [25, np.nan, 38, 52, np.nan, 41],
    "cholesterol": [180, 210, np.nan, 260, 240, np.nan],
    "smoker": ["No", "Yes", "No", np.nan, "No", "Yes"]
}

patients = pd.DataFrame(raw_records)
print("Missing Count Per Column:")
print(patients.isnull().sum())

# 1. Median Imputation for Age
median_age = patients["age"].median()
patients["age_clean"] = patients["age"].fillna(median_age)

# 2. Mean Imputation for Cholesterol
mean_chol = patients["cholesterol"].mean()
patients["cholesterol_clean"] = patients["cholesterol"].fillna(mean_chol)

# 3. Mode Imputation for Categorical (Smoker)
mode_smoker = patients["smoker"].mode()[0]
patients["smoker_clean"] = patients["smoker"].fillna(mode_smoker)

print("\nCleaned Dataset:")
print(patients[["patient_id", "age_clean", "cholesterol_clean", "smoker_clean"]])
```
