4. **The Scaled Dot-Product Attention Formula:**
   $$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{Q K^T}{\sqrt{d_k}}\right) V$$
5. **Why Scale by $\sqrt{d_k}$? (Preventing Tiny Gradients)**
6. **Hands-On NumPy Implementation from Scratch**

```python
import numpy as np

def softmax(x, axis=-1):
    e_x = np.exp(x - np.max(x, axis=axis, keepdims=True))
    return e_x / np.sum(e_x, axis=axis, keepdims=True)

def scaled_dot_product_attention(Q, K, V):
    d_k = Q.shape[-1]
    # 1. Compute dot product similarity: Q @ K^T
    scores = np.matmul(Q, K.T) / np.sqrt(d_k)
    # 2. Compute Attention Weights (Probabilities)
    weights = softmax(scores, axis=-1)
    # 3. Weighted sum of Values: weights @ V
    output = np.matmul(weights, V)
    return output, weights

# Sequence of 3 word tokens, each with dimension d=4: ["The", "bank", "river"]
tokens = np.array([
    [1.0, 0.5, 0.2, 0.0],  # "The"
    [0.1, 0.9, 0.8, 0.2],  # "bank"
    [0.0, 0.8, 0.9, 0.1]   # "river"
])

# For self-attention, Q, K, V are linear projections of the tokens
Q = tokens @ np.eye(4)
K = tokens @ np.eye(4)
V = tokens @ np.eye(4)

attn_out, attn_weights = scaled_dot_product_attention(Q, K, V)
print("=== Attention Weight Matrix (Who is attending to whom?) ===")
print("Row=Query Word, Col=Key Word:")
print(attn_weights.round(3))
print("\nNotice: 'bank' (Row 1) pays high attention to 'river' (Col 2) -> Disambiguating river bank vs money bank!")
```
