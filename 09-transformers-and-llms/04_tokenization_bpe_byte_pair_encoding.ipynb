```python
from collections import Counter

# Initial corpus with frequency of words
corpus = {"l o w </w>": 5, "l o w e r </w>": 2, "n e w e s t </w>": 6, "w i d e s t </w>": 3}

def get_stats(corpus):
    pairs = Counter()
    for word, freq in corpus.items():
        symbols = word.split()
        for i in range(len(symbols) - 1):
            pairs[symbols[i], symbols[i + 1]] += freq
    return pairs

def merge_vocab(pair, corpus):
    v_out = {}
    bigram = " ".join(pair)
    replacement = "".join(pair)
    for word in corpus:
        w_out = word.replace(bigram, replacement)
        v_out[w_out] = corpus[word]
    return v_out

# Run 3 BPE merge rounds
print("Initial Corpus:", corpus)
for round_num in range(3):
    pairs = get_stats(corpus)
    best_pair = max(pairs, key=pairs.get)
    corpus = merge_vocab(best_pair, corpus)
    print(f"Round {round_num + 1}: Merged {best_pair} (Occurrences: {pairs[best_pair]}) -> New Corpus: {corpus}")
```
