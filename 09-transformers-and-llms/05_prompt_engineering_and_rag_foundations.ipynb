```python
# Demonstrating In-Memory Vector Search Retrieval for RAG
import numpy as np

# Knowledge Base Chunks
knowledge_chunks = [
    "Antigravity CLI provides automated multi-agent coordination.",
    "Python 3.11 introduced substantial performance enhancements and faster exception handling.",
    "The transformer architecture was proposed by Vaswani et al. in 2017."
]

# Simulated 3D embeddings
chunk_embeddings = np.array([
    [0.9, 0.1, 0.0],  # Antigravity CLI
    [0.1, 0.8, 0.3],  # Python 3.11
    [0.2, 0.1, 0.9]   # Transformers
])

# User query: "Who invented transformers?" -> Embedded to [0.22, 0.08, 0.94]
query_embedding = np.array([0.22, 0.08, 0.94])

# Cosine similarity matching
similarities = np.dot(chunk_embeddings, query_embedding) / (
    np.linalg.norm(chunk_embeddings, axis=1) * np.linalg.norm(query_embedding)
)

best_chunk_idx = np.argmax(similarities)
print(f"Query: 'Who invented transformers?'")
print(f"Retrieved Top Chunk (Score: {similarities[best_chunk_idx]:.3f}):")
print(f"--> \"{knowledge_chunks[best_chunk_idx]}\"")
```
