```python
import numpy as np

# Layer Normalization from Scratch (Normalizes across features for EACH sample)
def layer_norm(x, eps=1e-5):
    mean = np.mean(x, axis=-1, keepdims=True)
    std = np.std(x, axis=-1, keepdims=True)
    return (x - mean) / (std + eps)

sample_token_embedding = np.array([[1.5, 3.2, -0.8, 5.0]])
normalized = layer_norm(sample_token_embedding)

print("Original Embedding :", sample_token_embedding)
print("Layer Normalized   :", normalized.round(3))
print("Mean (Target ~0.0) :", round(np.mean(normalized), 5))
print("Std  (Target ~1.0) :", round(np.std(normalized), 5))
```
