```python
import numpy as np

# Sinusoidal Positional Encoding implementation
def get_positional_encoding(seq_len=5, d_model=4):
    pe = np.zeros((seq_len, d_model))
    for pos in range(seq_len):
        for i in range(0, d_model, 2):
            denom = np.power(10000, (2 * i) / d_model)
            pe[pos, i] = np.sin(pos / denom)
            pe[pos, i + 1] = np.cos(pos / denom)
    return pe

pe_matrix = get_positional_encoding(seq_len=4, d_model=4)
print("=== Sinusoidal Positional Encoding Matrix (4 Tokens x 4 Dim) ===")
print(pe_matrix.round(3))
print("\nAdded directly to input embeddings so the model knows word order!")
```
