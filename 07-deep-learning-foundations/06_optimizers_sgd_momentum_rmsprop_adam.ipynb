## 💡 1. The Evolution of Optimizers
- **Batch Gradient Descent:** Computes gradient over ALL 1,000,000 rows before taking 1 step. Extremely slow and RAM-heavy.
- **Mini-Batch SGD:** Computes gradient over small batches (32, 64, or 128 rows). Fast and noisy, which helps jump out of local minima!
- **Momentum:** Adds fraction $\beta_1$ of previous step direction to smooth out high-frequency oscillations.
- **RMSProp:** Divides learning rate by root of exponentially decaying average of squared gradients $\sqrt{v_t + \epsilon}$. Speeds up flat directions, slows down steep walls.
- **Adam:** Combines **Momentum** (1st moment $m_t$) + **RMSProp** (2nd moment $v_t$) with bias correction! Used in 95% of real-world AI projects.

```python
import numpy as np

# Adam Optimizer step implementation from scratch
class AdamOptimizer:
    def __init__(self, lr=0.001, beta1=0.9, beta2=0.999, eps=1e-8):
        self.lr = lr
        self.beta1 = beta1
        self.beta2 = beta2
        self.eps = eps
        self.m = 0.0  # 1st moment vector (Momentum)
        self.v = 0.0  # 2nd moment vector (RMSProp)
        self.t = 0    # Timestep counter

    def update(self, w, grad):
        self.t += 1
        # Update biased 1st moment estimate
        self.m = self.beta1 * self.m + (1 - self.beta1) * grad
        # Update biased 2nd raw moment estimate
        self.v = self.beta2 * self.v + (1 - self.beta2) * (grad ** 2)
        # Compute bias-corrected 1st moment
        m_hat = self.m / (1 - self.beta1 ** self.t)
        # Compute bias-corrected 2nd raw moment
        v_hat = self.v / (1 - self.beta2 ** self.t)
        # Update parameter
        w -= (self.lr * m_hat) / (np.sqrt(v_hat) + self.eps)
        return w

# Test Adam step on weight w=10.0 with gradient=2.0
opt = AdamOptimizer(lr=0.1)
w = 10.0
print("Initial Weight:", w)
for step in range(5):
    w = opt.update(w, grad=2.0)
    print(f"Step {step+1}: Updated Weight = {w:.4f}")
```
