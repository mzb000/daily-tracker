## 📐 1. Forward Propagation Equations
For a 2-layer network with input $\mathbf{X}$ $(N \times D)$:
- **Layer 1 (Hidden):**
  $$\mathbf{Z}^{[1]} = \mathbf{X} \mathbf{W}^{[1]} + \mathbf{b}^{[1]}$$
  $$\mathbf{A}^{[1]} = \text{ReLU}(\mathbf{Z}^{[1]})$$
- **Layer 2 (Output):**
  $$\mathbf{Z}^{[2]} = \mathbf{A}^{[1]} \mathbf{W}^{[2]} + \mathbf{b}^{[2]}$$
  $$\hat{\mathbf{Y}} = \sigma(\mathbf{Z}^{[2]})$$

```python
import numpy as np

# Seed for 100% reproducible weights
np.random.seed(42)

# Batch of 3 samples, each having 4 features (N=3, D=4)
X_batch = np.array([
    [1.0, 2.0, 0.5, -1.0],
    [0.0, 1.5, -2.0, 0.5],
    [-1.0, -0.5, 1.0, 2.0]
])

# Architecture dimensions: 4 Inputs -> 5 Hidden Neurons -> 1 Output Neuron
n_inputs = 4
n_hidden = 5
n_outputs = 1

# He Initialization for ReLU weights: W ~ N(0, sqrt(2/n_in))
W1 = np.random.randn(n_inputs, n_hidden) * np.sqrt(2.0 / n_inputs)
b1 = np.zeros((1, n_hidden))

# Xavier Initialization for Sigmoid output layer
W2 = np.random.randn(n_hidden, n_outputs) * np.sqrt(1.0 / n_hidden)
b2 = np.zeros((1, n_outputs))

# --- FORWARD PROPAGATION ---
# Layer 1: Linear + ReLU
Z1 = np.dot(X_batch, W1) + b1
A1 = np.maximum(0, Z1)

# Layer 2: Linear + Sigmoid
Z2 = np.dot(A1, W2) + b2
A2 = 1.0 / (1.0 + np.exp(-np.clip(Z2, -500, 500)))

print("=== Forward Pass Output Shapes ===")
print("Input X Shape     :", X_batch.shape) # (3, 4)
print("Hidden Layer 1 (A1):", A1.shape)      # (3, 5)
print("Output Layer (A2) :", A2.shape)      # (3, 1)
print("\nPredicted Probabilities for 3 samples:\n", A2.round(4))
```

### 🔍 Line-by-Line Code Breakdown:
- 'X_batch' (3 x 4): 3 training examples, each with 4 measurements.
- 'W1' (4 x 5): Maps 4 inputs to 5 hidden neurons.
- 'np.dot(X_batch, W1) + b1': Matrix multiplication. The bias 'b1' (1 x 5) broadcasts across all 3 sample rows.
- 'A2' (3 x 1): Final sigmoid probability for each of the 3 samples.
