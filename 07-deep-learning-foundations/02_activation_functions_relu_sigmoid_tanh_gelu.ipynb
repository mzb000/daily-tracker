## 💡 1. Why Activation Functions Are Mandatory
Suppose you build a 100-layer deep neural network without non-linear activation functions:
$$Z_1 = W_1 X + b_1$$
$$Z_2 = W_2 Z_1 + b_2 = W_2 (W_1 X + b_1) + b_2 = (W_2 W_1) X + (W_2 b_1 + b_2)$$
Notice: The product of two weight matrices $(W_2 W_1)$ is just **another single linear matrix $W_{new}$**!
No matter how many millions of linear layers you stack, the entire 100-layer network mathematically collapses into **a single linear regression line**!
Non-linear activation functions bend the feature space, allowing networks to approximate ANY complex continuous function (The Universal Approximation Theorem).

```python
import numpy as np

# 1. Sigmoid Function: Squashes to (0, 1)
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))

def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1.0 - s)

# 2. Tanh Function: Squashes to (-1, 1) [Zero-Centered]
def tanh(z):
    return np.tanh(z)

def tanh_derivative(z):
    return 1.0 - np.tanh(z)**2

# 3. ReLU (Rectified Linear Unit): max(0, z)
def relu(z):
    return np.maximum(0, z)

def relu_derivative(z):
    return np.where(z > 0, 1.0, 0.0)

# 4. GELU (Gaussian Error Linear Unit - Used in GPT-4 and BERT)
def gelu(z):
    return 0.5 * z * (1.0 + np.tanh(np.sqrt(2.0 / np.pi) * (z + 0.044715 * np.power(z, 3))))

# Test across input range from -4 to +4
z_vals = np.array([-4.0, -1.0, 0.0, 1.0, 4.0])
print("Input z Values :", z_vals)
print("Sigmoid(z)     :", sigmoid(z_vals).round(3))
print("Tanh(z)        :", tanh(z_vals).round(3))
print("ReLU(z)        :", relu(z_vals).round(3))
print("GELU(z)        :", gelu(z_vals).round(3))
```

## ⚠️ 2. The Vanishing Gradient Problem Explained
Look at the derivative of Sigmoid:
$$\sigma'(z) = \sigma(z)(1 - \sigma(z))$$
- When $z = 0$, $\sigma'(0) = 0.5 \times 0.5 = \mathbf{0.25}$ (Maximum possible gradient!).
- When $z = 5$ or $z = -5$, $\sigma'(z) \approx \mathbf{0.0001}$!
In a 10-layer network, backpropagation multiplies derivatives using the chain rule:
$$(0.25)^{10} \approx 0.00000095$$
The gradient shrinks exponentially until it becomes **effectively zero**! The front layers stop learning completely.
This is why **ReLU** took over the world: for positive numbers ($z > 0$), its derivative is always **$1.0$**, so gradients flow backwards through 100+ layers without vanishing!
