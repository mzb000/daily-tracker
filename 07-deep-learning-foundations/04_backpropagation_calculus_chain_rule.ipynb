## 💡 1. The Core Idea: The Calculus Chain Rule
During forward pass, we calculate the Loss $L$.
During backward pass, we want to know:
$$\frac{\partial L}{\partial W^{[1]}} = \text{How much did weight } W^{[1]} \text{ contribute to the final error?}$$
Using the chain rule:
$$\frac{\partial L}{\partial W^{[1]}} = \frac{\partial L}{\partial A^{[2]}} \cdot \frac{\partial A^{[2]}}{\partial Z^{[2]}} \cdot \frac{\partial Z^{[2]}}{\partial A^{[1]}} \cdot \frac{\partial A^{[1]}}{\partial Z^{[1]}} \cdot \frac{\partial Z^{[1]}}{\partial W^{[1]}}$$
Gradients flow backwards from the output layer to the input layer like ripples in water!

```python
import numpy as np

# Let's solve the historic XOR problem that defeated the single-layer perceptron!
X_xor = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_xor = np.array([[0], [1], [1], [0]])

np.random.seed(42)
n_in, n_hid, n_out = 2, 4, 1

# Initialize parameters
W1 = np.random.randn(n_in, n_hid) * 0.5
b1 = np.zeros((1, n_hid))
W2 = np.random.randn(n_hid, n_out) * 0.5
b2 = np.zeros((1, n_out))

lr = 0.5
epochs = 5000

for epoch in range(epochs):
    # 1. FORWARD PASS
    Z1 = np.dot(X_xor, W1) + b1
    A1 = np.tanh(Z1)  # Using Tanh for hidden layer
    Z2 = np.dot(A1, W2) + b2
    A2 = 1.0 / (1.0 + np.exp(-Z2))  # Sigmoid for output

    # 2. BACKWARD PASS (CHAIN RULE)
    # Binary Cross-Entropy derivative simplifies beautifully with Sigmoid to: (A2 - y)
    dZ2 = A2 - y_xor
    dW2 = np.dot(A1.T, dZ2) / len(X_xor)
    db2 = np.sum(dZ2, axis=0, keepdims=True) / len(X_xor)

    # Backprop into hidden layer
    dA1 = np.dot(dZ2, W2.T)
    dZ1 = dA1 * (1.0 - np.tanh(Z1)**2)  # Tanh derivative
    dW1 = np.dot(X_xor.T, dZ1) / len(X_xor)
    db1 = np.sum(dZ1, axis=0, keepdims=True) / len(X_xor)

    # 3. GRADIENT DESCENT UPDATE
    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

# Evaluate on XOR
print("=== XOR Problem Solved by Multi-Layer Neural Network! ===")
print("True Target Labels:\n", y_xor.flatten())
print("Model Probabilities:\n", A2.round(3).flatten())
print("Binary Predictions :\n", (A2 >= 0.5).astype(int).flatten(), "--> 100% ACCURACY!")
```
