## 💡 1. Inverted Dropout (Hinton et al. 2012)
Imagine an office where every employee assumes someone else will do the work (co-adaptation of neurons).
If you randomly send 50% of the employees home on any given day, **everyone is forced to learn how to do every task independently!**
- During **Training**: Randomly drop activations with probability $p$ and scale remaining activations by $\frac{1}{1-p}$ (Inverted Dropout).
- During **Inference/Test**: Keep all neurons active at 100% capacity without scaling!

```python
import numpy as np

# Inverted Dropout implementation in NumPy
def inverted_dropout_forward(A, drop_prob=0.5, training=True):
    if not training or drop_prob == 0:
        return A, None

    keep_prob = 1.0 - drop_prob
    # Generate binary mask: 1 with probability keep_prob, 0 otherwise
    mask = (np.random.rand(*A.shape) < keep_prob).astype(np.float32)
    # Scale by 1/keep_prob so test-time activations don't need adjustment!
    A_drop = (A * mask) / keep_prob
    return A_drop, mask

# Test layer activations of 4 neurons
np.random.seed(42)
layer_activations = np.array([2.5, 4.0, 1.2, 3.8])

dropped_acts, mask = inverted_dropout_forward(layer_activations, drop_prob=0.5, training=True)
print("Original Activations :", layer_activations)
print("Dropout Binary Mask  :", mask)
print("Activations (Scaled) :", dropped_acts)
print("Notice: Neurons with mask=0 were zeroed out, surviving neurons scaled by 1/0.5 = 2x!")
```
