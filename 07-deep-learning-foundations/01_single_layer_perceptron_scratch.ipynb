## 💡 1. Biological Neuron vs Artificial Neuron
In the human brain:
- **Dendrites** receive electrical signals from other neurons.
- The **Cell Body (Soma)** aggregates the incoming signals.
- If the aggregated charge exceeds a threshold, the **Axon** fires an impulse to neighboring neurons!

In computer science:
- **Inputs ($x_1, x_2, \dots, x_n$):** Feature signals.
- **Weights ($w_1, w_2, \dots, w_n$):** The strength / importance of each input connection.
- **Bias ($b$):** The baseline threshold required to trigger the neuron.
- **Activation Function:** Step function that converts weighted sum into output (0 or 1).
$$z = \sum_{i=1}^{n} w_i x_i + b = \mathbf{w}^T \mathbf{x} + b$$
$$y = \begin{cases} 1 & \text{if } z \ge 0 \\ 0 & \text{if } z < 0 \end{cases}$$

```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=20):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = 0.0

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for epoch in range(self.epochs):
            errors = 0
            for xi, target in zip(X, y):
                # 1. Forward Pass (Linear combination)
                linear_output = np.dot(xi, self.weights) + self.bias
                # 2. Step Activation Function (Threshold = 0)
                y_pred = 1 if linear_output >= 0 else 0

                # 3. Perceptron Update Rule: w = w + lr * (y - y_hat) * x
                update = self.lr * (target - y_pred)
                self.weights += update * xi
                self.bias += update

                if update != 0.0:
                    errors += 1

            if errors == 0:
                print(f"Converged cleanly at epoch {epoch+1}!")
                break

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return np.where(linear_output >= 0, 1, 0)

# Training Perceptron to solve the Boolean AND Gate
# AND Gate Truth Table: Output is 1 only when BOTH inputs are 1
X_and = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([0, 0, 0, 1])

clf = Perceptron(learning_rate=0.1, epochs=10)
clf.fit(X_and, y_and)

print("Trained Weights:", clf.weights)
print("Trained Bias   :", round(clf.bias, 2))
print("Predictions    :", clf.predict(X_and), "(Expected: [0, 0, 0, 1])")
```

## ⚠️ 2. The Historic XOR Problem (The AI Winter)
In 1969, Marvin Minsky and Seymour Papert proved that a single-layer perceptron can **ONLY learn linearly separable patterns**.
- **AND gate:** Linearly separable (a straight line can separate 0s from 1).
- **OR gate:** Linearly separable.
- **XOR gate (Exclusive OR):** Output is 1 if inputs are different ($[0,1] \to 1, [1,0] \to 1$), but 0 if inputs are identical ($[0,0] \to 0, [1,1] \to 0$).
There is **NO single straight line** on a 2D plane that can separate the 1s from the 0s in XOR!
This failure caused funding to dry up for 15 years ("The First AI Winter"), until the discovery of **Multi-Layer Perceptrons (MLP) with Backpropagation**!

## 🎯 3. Top Interview Questions on Perceptrons
### Q1: What guarantees that the Perceptron Learning Algorithm will converge?
- **Answer:** The **Perceptron Convergence Theorem** (Novikoff 1962). If the training dataset is strictly linearly separable with margin $\gamma > 0$, the perceptron is mathematically guaranteed to find a separating hyperplane in a finite number of steps ($k \le (R / \gamma)^2$).
- If data is NOT linearly separable, the weights will oscillate infinitely without converging!
