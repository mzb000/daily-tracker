## 💡 1. Why PyTorch Won the Deep Learning Race
NumPy is CPU-only and requires manual calculus for derivatives.
**PyTorch** provides two superpowers:
1. **GPU Acceleration:** Can run tensor operations on NVIDIA CUDA cores 50x-200x faster than CPU!
2. **Autograd:** Automatically constructs a Directed Acyclic Graph (DAG) during the forward pass and computes exact gradients with .backward()!

```python
import torch
import torch.nn as nn

# 1. Tensor creation and Autograd
x = torch.tensor([3.0], requires_grad=True)

# Define function: y = 2 * x^3 + 5
y = 2 * (x ** 3) + 5

# Backpropagation: Compute dy/dx = 6 * x^2
y.backward()

print("x value:", x.item())
print("y value:", y.item())
print("dy/dx gradient (Analytical: 6 * 3^2 = 54):", x.grad.item())
```

## 🏗️ 2. Building a Classifier with torch.nn.Module

```python
class SimpleClassifier(nn.Module):
    def __init__(self, input_dim=4, hidden_dim=8, output_dim=1):
        super(SimpleClassifier, self).__init__()
        self.network = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, output_dim),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.network(x)

# Instantiate model
model = SimpleClassifier()
print(model)

# Forward pass with random tensor
dummy_input = torch.randn(2, 4)
output_probs = model(dummy_input)
print("\nForward pass output shape:", output_probs.shape)
print("Output probabilities:\n", output_probs)
```
