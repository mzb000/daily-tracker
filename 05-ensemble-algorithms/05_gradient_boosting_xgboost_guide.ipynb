## 🎯 Fitting the Residuals
In Gradient Boosting, each new tree is trained to predict the **residual error ($y - \hat{y}$)** of the combined previous trees!
$$\hat{y}_{new} = \hat{y}_{old} + \eta \cdot \text{Tree}_{residual}(X)$$
Where $\eta$ is the shrinkage learning rate (e.g., 0.1).

```python
import numpy as np

y_actual = np.array([50.0, 75.0, 100.0, 120.0])
# Round 0: Initial baseline prediction is the average
y_pred_0 = np.mean(y_actual)
residuals = y_actual - y_pred_0

print("Actual Values   :", y_actual)
print("Initial Average :", y_pred_0)
print("Residual Errors for Tree 1 to learn:", residuals)
```
