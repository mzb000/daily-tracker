## 💡 1. Parameter vs Hyperparameter
- **Model Parameters:** Learned automatically from data during training (Weights $w$, Bias $b$, Split thresholds).
- **Hyperparameters:** Configured by the Data Scientist **before** training starts (max_depth, learning_rate, n_estimators, regularization_lambda).

```python
import numpy as np

# Demonstrating K-Fold Cross-Validation Split Indices from scratch
n_samples = 12
k_folds = 3
indices = np.arange(n_samples)

fold_size = n_samples // k_folds

print("=== 3-Fold Cross-Validation Splits ===")
for fold in range(k_folds):
    val_indices = indices[fold * fold_size : (fold + 1) * fold_size]
    train_indices = np.setdiff1d(indices, val_indices)
    print(f"Fold {fold+1} -> Train rows: {train_indices} | Validation rows: {val_indices}")
```
