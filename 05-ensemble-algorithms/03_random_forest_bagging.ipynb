## 🎯 Wisdom of the Crowds
A single Decision Tree has high variance and easily overfits.
A **Random Forest** trains 100 or 500 different trees and lets them **vote**!
- **Bootstrap Sampling**: Each tree is trained on a random sample of rows with replacement.
- **Feature Subsampling**: At each split, only a random subset $\sqrt{D}$ of features is considered.
Result: The trees become decorrelated, and averaging their predictions dramatically reduces variance!

```python
import numpy as np

# Simulating voting across 5 trees in a forest
tree_predictions = np.array([
    [1, 1, 0, 1, 1],  # 4 vote 1, 1 votes 0 -> Majority = 1
    [0, 0, 0, 1, 0],  # 4 vote 0, 1 votes 1 -> Majority = 0
    [1, 0, 1, 1, 1]   # 4 vote 1, 1 votes 0 -> Majority = 1
])

# Majority vote along columns
majority_vote = (np.mean(tree_predictions, axis=1) >= 0.5).astype(int)
print("Forest Majority Predictions:", majority_vote)
```
