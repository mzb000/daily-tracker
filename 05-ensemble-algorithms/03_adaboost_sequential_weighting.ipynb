## 💡 1. The Paradigm Shift: Learning From Past Mistakes
In **Bagging (Random Forest)**, all trees are trained in parallel, independently.
In **Boosting (AdaBoost, XGBoost)**, trees are trained **sequentially in a chain**:
1. Tree 1 makes predictions.
2. We identify the difficult samples that Tree 1 got **wrong**.
3. We **increase the weights** of those misclassified samples.
4. Tree 2 is trained with heavy penalty on those specific mistakes!
5. Tree 3 focuses on what Tree 1 and Tree 2 together missed!

## 📐 2. The Mathematical Weight Update Rule
For sample $i$ with current weight $w_i$:
- If classified **correctly**: Weight is scaled down by $e^{-\alpha}$.
- If classified **incorrectly**: Weight is scaled up by $e^{+\alpha}$.
Where $\alpha = \frac{1}{2} \ln\left(\frac{1 - \epsilon}{\epsilon}\right)$ is the tree's voting power based on its error rate $\epsilon$.

```python
import numpy as np

# 5 training samples, all start with equal weights: 1/5 = 0.20
n = 5
weights = np.ones(n) / n
print("Initial Equal Weights:", weights)

# Suppose Model 1 correctly classifies samples 0, 1, 2, 4 but FAILS on sample 3
y_true = np.array([1, 1, 1, -1, 1])
y_pred = np.array([1, 1, 1,  1, 1])  # Sample index 3 was misclassified!

is_error = (y_true != y_pred).astype(int)
epsilon = np.sum(weights * is_error)  # Weighted error rate
alpha = 0.5 * np.log((1.0 - epsilon) / (epsilon + 1e-10))  # Model voting weight

# Update weights
weights = weights * np.exp(alpha * (2 * is_error - 1))
weights = weights / np.sum(weights)  # Normalize to sum to 1.0

print(f"\nModel Error Rate (epsilon): {epsilon:.2f}")
print(f"Model Say/Power (alpha)  : {alpha:.2f}")
print("\nUpdated Sample Weights:")
for i, w in enumerate(weights):
    status = "WRONG! (Boosted)" if is_error[i] else "Correct (Reduced)"
    print(f"Sample {i}: Weight = {w:.3f} [{status}]")
```

### 🔍 Line-by-Line Code Breakdown:
- Notice how Sample #3's weight jumped from **0.20 to ~0.50**!
- In the next round, any tree that misclassifies Sample #3 will suffer a massive 50% loss penalty. It is forced to learn how to classify that hard sample correctly!
