## 🎯 How Boosting Works:
Unlike Random Forest (where trees train in parallel independently):
1. **Tree 1** trains on the dataset.
2. We identify which samples Tree 1 got **wrong**.
3. We increase the weights of those hard samples so **Tree 2 is forced to focus on the mistakes**!
4. Repeat for $M$ rounds.

```python
import numpy as np

# Sample weights start equal
n_samples = 4
weights = np.ones(n_samples) / n_samples
print("Initial Equal Weights:", weights)

# Suppose sample #2 was misclassified
misclassified_idx = 1
# Reward correctly classified (lower weight), penalize error (higher weight)
weights[misclassified_idx] *= 2.5
weights /= np.sum(weights)  # Normalize to sum to 1

print("Updated Sample Weights (Sample #2 boosted!):", weights.round(3))
```
