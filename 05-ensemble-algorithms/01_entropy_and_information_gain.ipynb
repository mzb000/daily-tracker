## 🎯 The Quest for Pure Nodes
When building a Decision Tree, the goal is to ask questions (e.g., "Is Age > 30?") that split mixed groups into pure groups.

### Two Purity Formulas:
1. **Shannon Entropy (Surprise/Disorder)**:
   $$H(S) = -\sum_{i=1}^{C} p_i \log_2(p_i)$$
   - Pure node (100% yes, 0% no): $H(S) = 0$ (Zero disorder).
   - Maximum chaos (50% yes, 50% no): $H(S) = 1.0$.

2. **Gini Impurity (Probability of Misclassification)**:
   $$Gini(S) = 1 - \sum_{i=1}^{C} p_i^2$$

```python
import numpy as np

def entropy(probabilities):
    p = np.array(probabilities)
    p = p[p > 0]  # Avoid log2(0)
    return -np.sum(p * np.log2(p))

def gini_impurity(probabilities):
    p = np.array(probabilities)
    return 1.0 - np.sum(p ** 2)

# Perfectly Pure Node (all 10 items belong to Class A)
pure = [1.0, 0.0]
# Totally Chaotic Node (5 items Class A, 5 items Class B)
mixed = [0.5, 0.5]

print("Pure Node  - Entropy:", round(entropy(pure), 3), "| Gini:", round(gini_impurity(pure), 3))
print("Mixed Node - Entropy:", round(entropy(mixed), 3), "| Gini:", round(gini_impurity(mixed), 3))
```
