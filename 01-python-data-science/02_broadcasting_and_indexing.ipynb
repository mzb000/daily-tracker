## 🎯 What is Broadcasting?
Broadcasting is one of NumPy's most powerful features. It allows mathematical operations (addition, subtraction, multiplication) between arrays of **different shapes** without copying memory!

Imagine you have a matrix of student marks for 3 subjects:
```
Math  Physics  Chemistry
[70,     80,       90]   (Student A)
[65,     75,       85]   (Student B)
```
Now suppose the teacher wants to add **5 grace marks to Math**, **2 to Physics**, and **0 to Chemistry**.
The grace marks vector is: `[5, 2, 0]` (Shape: 1x3).
NumPy automatically **broadcasts** this vector across both students!

---

## 📏 The Golden Broadcasting Rules
NumPy compares shapes from right to left (trailing dimensions):
Two dimensions are compatible if:
1. They are **equal**, OR
2. One of them is **1**.

### Step 1: Broadcasting in Action

````python
import numpy as np

# Student marks (Shape: 2 rows, 3 columns)
marks = np.array([
    [70, 80, 90],
    [65, 75, 85]
])

# Grace marks per subject (Shape: 3,)
grace = np.array([5, 2, 0])

# NumPy broadcasts (3,) to match (2, 3) seamlessly!
final_marks = marks + grace

print("Original Marks:\n", marks)
print("Grace Vector   :", grace)
print("\nFinal Marks (After Broadcasting):\n", final_marks)
````

### Step 2: Boolean Masking (Fast Filtering)
In Data Science, we rarely use `if` conditions in loops. We use **Boolean Masks** to filter millions of rows in milliseconds.

````python
# Customer transaction amounts in dollars
transactions = np.array([15, 120, 45, 350, 80, 500, 25, 95])

# Find all high-value transactions (> $100)
mask = transactions > 100
print("Boolean Mask Array:\n", mask)

# Apply mask to extract matching elements
high_value = transactions[mask]
print("\nHigh-Value Transactions (> $100):", high_value)
````

### 🔍 Line-by-Line Code Breakdown:
- `transactions > 100`: Evaluates the comparison on every element at once, producing an array of `[False, True, False, True...]`.
- `transactions[mask]`: NumPy picks only the elements where the mask is `True`.
- This operation runs at C-speed in memory, avoiding slow Python `for` loops!

### Step 3: Multi-Condition Filtering (`&` and `|`)
Suppose we want transactions between $40 and $200.

````python
# Note: Use & for AND, | for OR, and wrap conditions in parentheses ()
medium_range = transactions[(transactions >= 40) & (transactions <= 200)]
print("Transactions between $40 and $200:", medium_range)
````
