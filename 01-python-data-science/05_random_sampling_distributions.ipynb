## 🎯 Why is Reproducibility Critical?
When training machine learning models:
- Neural network weights are initialized randomly.
- Datasets are split into Train/Test sets randomly.
- Random Forests select feature subsets randomly.

If you don't fix the **Random Seed**, you will get different accuracy scores every single time you run the code! That makes debugging impossible. By setting a seed, you guarantee **100% reproducible results** across team members and cloud servers.

```python
import numpy as np

# Modern NumPy best practice: Use Generator with explicit seed
rng = np.random.default_rng(seed=42)

# 1. Normal (Gaussian) Distribution (Mean=0, Std=1)
# Used in weight initialization and simulating natural phenomena
gaussian_samples = rng.normal(loc=0.0, scale=1.0, size=1000)

print("Gaussian Samples Summary:")
print(f"Sample Mean (Target ~0.0): {np.mean(gaussian_samples):.3f}")
print(f"Sample Std  (Target ~1.0): {np.std(gaussian_samples):.3f}")

# 2. Uniform Distribution (Every number between low and high is equally likely)
uniform_samples = rng.uniform(low=10.0, high=50.0, size=5)
print("\nUniform Samples between 10 and 50:\n", np.round(uniform_samples, 2))

# 3. Random Train/Test Index Shuffling
indices = np.arange(10)
print("\nOriginal Indices:", indices)
rng.shuffle(indices)
print("Shuffled Indices:", indices)
```
