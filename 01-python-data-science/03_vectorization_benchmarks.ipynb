## 🎯 The Core Philosophy: "No Python Loops Over Data"
Python is an interpreted language. When you write:
```python
for x in my_list:
    result.append(x * 2)
```
For EVERY SINGLE element, Python must:
1. Lookup the multiply method `__mul__`.
2. Check if `x` is an int, float, or string.
3. Allocate a new Python object on the heap.
4. Append the pointer to the list.

With 10,000,000 rows, this overhead takes seconds or minutes.
With **NumPy vectorization**, the calculation is delegated to pre-compiled C/Fortran routines that run directly in CPU registers!

````python
import time
import numpy as np

N = 2_000_000  # 2 Million numbers

# Setup pure python list and numpy array
py_list = list(range(N))
np_array = np.arange(N, dtype=np.float64)

# 1. Pure Python Loop
start_py = time.perf_counter()
py_result = [x * 3.5 for x in py_list]
elapsed_py = time.perf_counter() - start_py

# 2. NumPy Vectorized Operation
start_np = time.perf_counter()
np_result = np_array * 3.5
elapsed_np = time.perf_counter() - start_np

print(f"Dataset Size       : {N:,} elements")
print(f"Python Loop Time   : {elapsed_py:.4f} seconds")
print(f"NumPy Time         : {elapsed_np:.4f} seconds")
print(f"Speedup Factor     : {elapsed_py / elapsed_np:.1f}x faster with NumPy! 🚀")
````

### 📊 What Happened Under the Hood?
- **Python Loop**: Spent 90% of its time on object boxing/unboxing and interpreter loop overhead.
- **NumPy**: Loaded blocks of 64-bit numbers directly into SIMD (Single Instruction Multiple Data) registers (AVX2/AVX-512) and executed vector multiplication in raw assembly!
