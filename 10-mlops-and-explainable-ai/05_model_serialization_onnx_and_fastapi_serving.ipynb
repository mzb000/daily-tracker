```python
# FastAPI Service Template with Pydantic Schema Validation
sample_fastapi_code = """
from fastapi import FastAPI
from pydantic import BaseModel, Field

app = FastAPI(title="Credit Scoring ML API", version="1.0.0")

class LoanApplication(BaseModel):
    income: float = Field(..., gt=0, description="Annual income in USD")
    credit_score: int = Field(..., ge=300, le=850, description="FICO score")
    debt_to_income: float = Field(..., ge=0.0, le=1.0)

@app.post("/predict")
def predict_default(app: LoanApplication):
    # Production model inference logic here
    prob_default = 0.05 if app.credit_score > 700 else 0.45
    return {"probability_of_default": prob_default, "decision": "Approve" if prob_default < 0.20 else "Reject"}
"""
print("Production FastAPI Serving Template:")
print(sample_fastapi_code)
```
