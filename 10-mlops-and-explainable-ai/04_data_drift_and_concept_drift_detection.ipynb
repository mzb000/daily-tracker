```python
import numpy as np

# Population Stability Index (PSI) implementation from scratch
def calculate_psi(baseline, production, num_buckets=5):
    # Quantile bin edges on baseline
    percentiles = np.linspace(0, 100, num_buckets + 1)
    bins = np.percentile(baseline, percentiles)
    bins[0] -= 1e-5
    bins[-1] += 1e-5

    base_counts, _ = np.histogram(baseline, bins=bins)
    prod_counts, _ = np.histogram(production, bins=bins)

    base_pct = base_counts / len(baseline) + 1e-4
    prod_pct = prod_counts / len(production) + 1e-4

    psi_val = np.sum((prod_pct - base_pct) * np.log(prod_pct / base_pct))
    return psi_val

np.random.seed(42)
baseline_data = np.random.normal(50, 10, 1000)
prod_data_normal = np.random.normal(50, 10, 1000)
prod_data_drifted = np.random.normal(65, 12, 1000)  # Significant shift!

print("PSI on Stable Data :", round(calculate_psi(baseline_data, prod_data_normal), 4), "(< 0.10: No Drift)")
print("PSI on Drifted Data:", round(calculate_psi(baseline_data, prod_data_drifted), 4), "(> 0.25: Severe Drift! Trigger Retraining)")
```
