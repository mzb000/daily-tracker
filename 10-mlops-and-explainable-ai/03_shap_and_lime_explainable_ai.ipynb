```python
import numpy as np

# Intuition of Shapley Values: 2 Features (Experience, Degree) predicting salary boost
# Baseline salary: $50k
# With Experience only: $70k (+20k)
# With Degree only: $65k (+15k)
# With Both: $90k (+40k)

# Shapley calculation averages marginal contributions across permutations:
# Order 1 (Exp then Deg): Exp gets +20k, Deg gets (+40k - 20k) = +20k
# Order 2 (Deg then Exp): Deg gets +15k, Exp gets (+40k - 15k) = +25k

shap_experience = (20 + 25) / 2
shap_degree = (20 + 15) / 2

print("=== Shapley Feature Attributions ===")
print("Baseline Salary     : USD 50k")
print(f"Experience Impact   : +USD {shap_experience}k")
print(f"Degree Impact       : +USD {shap_degree}k")
print(f"Final Salary Sum    : USD {50 + shap_experience + shap_degree}k (Exact Additive Consistency!)")
```
