```python
import numpy as np

# Evaluating False Positive vs False Negative Business Costs
# FP Cost: $5 (Customer annoyed by false fraud alert)
# FN Cost: $500 (Actual fraudster stole money!)

probabilities = np.array([0.15, 0.35, 0.45, 0.65, 0.85])
actual_labels = np.array([0,    0,    1,    1,    1])

thresholds = [0.3, 0.5, 0.7]
print("=== Business Cost Optimization Across Decision Thresholds ===")
for th in thresholds:
    preds = (probabilities >= th).astype(int)
    fp = np.sum((actual_labels == 0) & (preds == 1))
    fn = np.sum((actual_labels == 1) & (preds == 0))
    total_cost = (fp * 5) + (fn * 500)
    print(f"Threshold {th:.1f} -> FP: {fp}, FN: {fn} -> Total Financial Cost: USD {total_cost}")
```
