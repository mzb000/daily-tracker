```python
import numpy as np

# Time Series Expanding Window Cross-Validation from scratch
n_timesteps = 8
min_train_size = 3

print("=== Time Series Expanding Window Validation ===")
for split_idx, end_train in enumerate(range(min_train_size, n_timesteps)):
    train_idx = list(range(0, end_train))
    val_idx = [end_train]
    print(f"Split {split_idx + 1} -> Train Days: {train_idx} | Test Day: {val_idx} (Never train on the future!)")
```
