```python
import numpy as np

# Vanilla RNN Forward Step in Pure NumPy
class SimpleRNNCell:
    def __init__(self, input_dim=3, hidden_dim=4):
        self.W_xh = np.random.randn(input_dim, hidden_dim) * 0.1
        self.W_hh = np.random.randn(hidden_dim, hidden_dim) * 0.1
        self.b_h  = np.zeros((1, hidden_dim))

    def step(self, x_t, h_prev):
        # h_t = tanh(x_t @ W_xh + h_prev @ W_hh + b_h)
        h_next = np.tanh(np.dot(x_t, self.W_xh) + np.dot(h_prev, self.W_hh) + self.b_h)
        return h_next

rnn = SimpleRNNCell(input_dim=3, hidden_dim=4)
h = np.zeros((1, 4))  # Initial zero hidden state

# Process sequence of 3 timesteps: x1, x2, x3
sequence = [np.random.randn(1, 3) for _ in range(3)]
for t, x_t in enumerate(sequence):
    h = rnn.step(x_t, h)
    print(f"Timestep {t+1} -> Hidden Memory State (h_{t+1}):\n", h.round(3))
```
