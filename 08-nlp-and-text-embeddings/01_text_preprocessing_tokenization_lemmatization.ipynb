```python
import re

raw_text = "Hello! Check out https://github.com/mzb000/daily-tracker #AI. The developers are running 100+ experiments!"

# 1. Clean URLs and Special Characters
cleaned_text = re.sub(r"https?://\S+", "", raw_text)
cleaned_text = re.sub(r"[^a-zA-Z0-9\s]", "", cleaned_text).lower()

# 2. Tokenize by Whitespace
tokens = cleaned_text.split()

# 3. Stop Word Filtering (Common grammatical glue words)
stop_words = {"the", "are", "out", "a", "an", "in", "on", "at"}
filtered_tokens = [w for w in tokens if w not in stop_words]

print("Original Text   :", raw_text)
print("Cleaned Tokens  :", tokens)
print("Filtered Tokens :", filtered_tokens)
```
