```python
import numpy as np

def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

# Simulated 3D semantic word vectors: [Royalty, Masculinity, Femininity]
v_king   = np.array([0.95,  0.88,  0.02])
v_man    = np.array([0.05,  0.92,  0.01])
v_woman  = np.array([0.04,  0.02,  0.95])
v_queen  = np.array([0.96,  0.03,  0.92])
v_apple  = np.array([-0.5, -0.4, -0.3])

# Vector Arithmetic: King - Man + Woman ~ Queen
v_analogy = v_king - v_man + v_woman

print("Cosine Similarity(King, Queen) :", round(cosine_similarity(v_king, v_queen), 3))
print("Cosine Similarity(King, Apple) :", round(cosine_similarity(v_king, v_apple), 3))
print("Cosine Similarity(Analogy, Queen):", round(cosine_similarity(v_analogy, v_queen), 3), "(Near Perfect Match!)")
```
