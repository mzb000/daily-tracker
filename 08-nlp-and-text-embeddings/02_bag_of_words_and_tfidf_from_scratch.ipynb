```python
import numpy as np

documents = [
    "machine learning is amazing",
    "deep learning is powerful",
    "machine learning and deep learning"
]

# Build Vocabulary
words = sorted(list(set(" ".join(documents).split())))
word_to_id = {w: i for i, w in enumerate(words)}
n_docs = len(documents)
n_vocab = len(words)

# Compute Term Frequency (TF)
tf = np.zeros((n_docs, n_vocab))
for doc_id, doc in enumerate(documents):
    tokens = doc.split()
    for t in tokens:
        tf[doc_id, word_to_id[t]] += 1
    tf[doc_id] /= len(tokens)

# Compute Inverse Document Frequency (IDF)
doc_counts = np.sum(tf > 0, axis=0)
idf = np.log((n_docs + 1) / (doc_counts + 1)) + 1.0  # Smooth IDF

tfidf = tf * idf
# L2 Normalize
tfidf /= np.linalg.norm(tfidf, axis=1, keepdims=True)

print("Vocabulary Words:", words)
print("\nTF-IDF Matrix Shape:", tfidf.shape)
print("Doc 0 TF-IDF Vector:\n", tfidf[0].round(3))
```
