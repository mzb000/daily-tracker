```python
import numpy as np

# Training messages: [Message, Label: 1=Spam, 0=Ham]
train_data = [
    ("win free cash now", 1),
    ("urgent lottery winner claim cash", 1),
    ("are we meeting for coffee today", 0),
    ("project report deadline extended", 0)
]

# Vocabulary building
spam_words = {}
ham_words = {}
total_spam, total_ham = 0, 0

for msg, label in train_data:
    words = msg.split()
    if label == 1:
        total_spam += len(words)
        for w in words: spam_words[w] = spam_words.get(w, 0) + 1
    else:
        total_ham += len(words)
        for w in words: ham_words[w] = ham_words.get(w, 0) + 1

vocab_size = len(set(list(spam_words.keys()) + list(ham_words.keys())))

# Classify new unseen message: "claim free cash"
test_msg = "claim free cash"
p_spam = np.log(2/4)  # Prior
p_ham  = np.log(2/4)

for w in test_msg.split():
    # Laplace smoothing: (count + 1) / (total + vocab_size)
    p_w_given_spam = (spam_words.get(w, 0) + 1) / (total_spam + vocab_size)
    p_w_given_ham  = (ham_words.get(w, 0) + 1) / (total_ham + vocab_size)
    p_spam += np.log(p_w_given_spam)
    p_ham  += np.log(p_w_given_ham)

print(f"Log-Probability Spam: {p_spam:.3f}")
print(f"Log-Probability Ham : {p_ham:.3f}")
print("Predicted Class     :", "SPAM" if p_spam > p_ham else "HAM")
```
