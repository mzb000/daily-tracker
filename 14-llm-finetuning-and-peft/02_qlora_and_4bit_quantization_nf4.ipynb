```python
# Memory footprint calculation for 7B, 13B, and 70B parameter models
def calculate_vram(params_billions):
    fp16_gb = params_billions * 2  # 2 bytes per weight
    int4_gb = params_billions * 0.5 # 0.5 bytes per weight (4 bits)
    return fp16_gb, int4_gb

print("=== Model VRAM Footprint: FP16 vs 4-Bit Quantization ===")
for p in [7, 13, 70]:
    fp16, q4 = calculate_vram(p)
    print(f"{p:2d}B Model -> Full FP16: {fp16:3d} GB VRAM | 4-Bit QLoRA: {q4:5.1f} GB VRAM (Fits easily on consumer GPUs!)")
```
