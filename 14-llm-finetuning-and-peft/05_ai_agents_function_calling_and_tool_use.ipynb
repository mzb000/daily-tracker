```python
import json

# Defining a Function Calling Tool Schema
calculator_tool = {
    "name": "calculate_mortgage",
    "description": "Calculates monthly mortgage payment given loan, interest rate, and term.",
    "parameters": {
        "type": "object",
        "properties": {
            "loan_amount": {"type": "number", "description": "Principal amount in USD"},
            "interest_rate": {"type": "number", "description": "Annual interest rate (e.g. 0.065)"},
            "years": {"type": "integer", "description": "Loan term in years"}
        },
        "required": ["loan_amount", "interest_rate", "years"]
    }
}

# The ReAct Agent Loop Simulation
user_prompt = "What is the monthly payment on a 300,000 dollar loan at 6% over 30 years?"
thought = "I need to call the calculate_mortgage tool with loan_amount=300000, interest_rate=0.06, years=30."
action_call = {"tool": "calculate_mortgage", "arguments": {"loan_amount": 300000, "interest_rate": 0.06, "years": 30}}

# Tool execution (Observation)
P = 300000
r = 0.06 / 12
n = 30 * 12
monthly_payment = (P * r * (1 + r)**n) / ((1 + r)**n - 1)

print("User Prompt :", user_prompt)
print("Thought     :", thought)
print("Action Call :", json.dumps(action_call, indent=2))
print(f"Observation : Tool output = USD {monthly_payment:.2f} per month")
print(f"Final Answer: Your estimated monthly mortgage payment is USD {monthly_payment:.2f}!")
```
