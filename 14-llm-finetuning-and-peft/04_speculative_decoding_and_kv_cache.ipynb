```python
import numpy as np

# Simulating KV-Cache Growth across 4 generated tokens
kv_cache = {"keys": [], "values": []}

def generate_next_token(new_token_id, step):
    # Generate new key & value vectors for the single new token
    k = np.array([float(new_token_id), step])
    v = np.array([step * 1.5, float(new_token_id) * 0.5])
    
    # Append to cache
    kv_cache["keys"].append(k)
    kv_cache["values"].append(v)
    
    # Total cached tokens available for attention
    cached_len = len(kv_cache["keys"])
    return cached_len

for t in range(4):
    total_cached = generate_next_token(new_token_id=100 + t, step=t)
    print(f"Token {t+1} generated -> Total Keys in Cache: {total_cached} (No recomputing past tokens!)")
```
