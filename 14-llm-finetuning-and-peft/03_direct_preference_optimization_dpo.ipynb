3. **The DPO Loss Formula:**
   $$\mathcal{L}_{\text{DPO}}(\pi_\theta) = -\mathbb{E} \left[ \log \sigma \left( \beta \log \frac{\pi_\theta(y_w | x)}{\pi_{\text{ref}}(y_w | x)} - \beta \log \frac{\pi_\theta(y_l | x)}{\pi_{\text{ref}}(y_l | x)} \right) \right]$$
4. **Hands-On DPO Loss Implementation in Pure NumPy**

```python
import numpy as np

def dpo_loss(policy_win_logp, policy_lose_logp, ref_win_logp, ref_lose_logp, beta=0.1):
    # Log-ratio for winning answer
    log_ratio_win = policy_win_logp - ref_win_logp
    # Log-ratio for losing answer
    log_ratio_lose = policy_lose_logp - ref_lose_logp
    # Implicit reward difference
    implicit_reward_diff = beta * (log_ratio_win - log_ratio_lose)
    loss = -np.log(1.0 / (1.0 + np.exp(-implicit_reward_diff)))
    return loss

# Suppose winning answer probability increased in our fine-tuned policy
loss = dpo_loss(
    policy_win_logp=-1.2, policy_lose_logp=-4.5,
    ref_win_logp=-2.0,    ref_lose_logp=-2.1,
    beta=0.1
)
print(f"Computed DPO Loss: {loss:.4f} (Directly aligns policy without a reward model!)")
```
