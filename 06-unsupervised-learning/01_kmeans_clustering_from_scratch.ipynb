## 💡 1. Unsupervised Learning: Customer Segmentation
In Supervised Learning, we had labels $y$ (e.g., Churned = 0 or 1).
In **Unsupervised Learning**, we only have features $X$!
The goal is to discover **hidden natural clusters**:
- High-spending frequent shoppers vs Bargain hunters.
- Fraud ring networks vs Normal card transactions.

## 🔄 2. The 2-Step K-Means Loop (Lloyd's Algorithm)
1. **Assignment Step:** Measure Euclidean distance from every point to all $K$ centroids. Assign each point to its nearest centroid.
2. **Update Step:** Recalculate each centroid as the **mean** of all points currently assigned to it!
Repeat until centroids stop moving (convergence).

```python
import numpy as np

class KMeansScratch:
    def __init__(self, k=2, max_iters=100):
        self.k = k
        self.max_iters = max_iters
        self.centroids = None

    def fit(self, X):
        # 1. Randomly pick k initial centroids from the dataset
        np.random.seed(42)
        random_indices = np.random.choice(len(X), size=self.k, replace=False)
        self.centroids = X[random_indices]

        for _ in range(self.max_iters):
            # Distance from all points to all centroids: Shape (N, K)
            distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2)
            # Assign points to closest centroid
            labels = np.argmin(distances, axis=1)

            # Recompute centroids as mean of assigned points
            new_centroids = np.array([X[labels == i].mean(axis=0) for i in range(self.k)])

            # Check convergence
            if np.allclose(self.centroids, new_centroids):
                break
            self.centroids = new_centroids

        return labels

# Synthetic 2D Customer Data: [Annual Spend ($k), Visit Frequency]
customers = np.array([
    [15, 2], [18, 3], [12, 1],   # Cluster 0: Low spenders
    [85, 12], [90, 15], [80, 10]  # Cluster 1: VIP frequent buyers
], dtype=np.float64)

kmeans = KMeansScratch(k=2)
cluster_assignments = kmeans.fit(customers)

print("Customer Cluster Assignments:", cluster_assignments)
print("\nFinal Cluster Centroids:\n", kmeans.centroids.round(2))
```

### 🔍 Line-by-Line Code Breakdown:
- 'X[:, np.newaxis] - self.centroids': Uses NumPy broadcasting to compute pairwise differences between N data points and K centroids simultaneously in memory without loops!
- 'np.argmin(distances, axis=1)': For each row, picks the index (0 or 1) of the minimum distance centroid.
- 'np.allclose(self.centroids, new_centroids)': Stops training as soon as centroids freeze into stable coordinates.
