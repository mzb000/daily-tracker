## 💡 1. The Inertia Curve & The Elbow
As $K$ increases, Inertia (distance from points to their centroid) naturally drops. If $K = N$, Inertia is $0$, but that is useless!
We look for the **Elbow Point**—the point where adding another cluster gives diminishing returns in variance explained.

```python
import numpy as np

# Calculating Within-Cluster Sum of Squares (Inertia)
def compute_inertia(X, labels, centroids):
    inertia = 0.0
    for k in range(len(centroids)):
        cluster_points = X[labels == k]
        if len(cluster_points) > 0:
            inertia += np.sum((cluster_points - centroids[k]) ** 2)
    return inertia

sample_data = np.array([[1, 2], [1, 4], [1, 0], [10, 2], [10, 4], [10, 0]])
centroids_k1 = np.array([[5.5, 2.0]])
labels_k1 = np.zeros(len(sample_data))

centroids_k2 = np.array([[1.0, 2.0], [10.0, 2.0]])
labels_k2 = np.array([0, 0, 0, 1, 1, 1])

print(f"Inertia for K=1: {compute_inertia(sample_data, labels_k1, centroids_k1):.2f}")
print(f"Inertia for K=2: {compute_inertia(sample_data, labels_k2, centroids_k2):.2f} (Massive 80% reduction -> K=2 is the Elbow!)")
```
