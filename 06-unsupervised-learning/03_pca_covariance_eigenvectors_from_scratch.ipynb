3. **The 5-Step PCA Algorithm:**
   - Step 1: Standardize the features (Mean = 0).
   - Step 2: Compute the Covariance Matrix $\Sigma = \frac{1}{N} X^T X$.
   - Step 3: Compute Eigenvectors and Eigenvalues.
   - Step 4: Sort Eigenvectors by descending Eigenvalue magnitude.
   - Step 5: Project high-D data onto the top $K$ Principal Components!

```python
import numpy as np

# 4 samples with 3 features: [Feature A, Feature B, Feature C]
X = np.array([
    [2.5, 2.4, 0.5],
    [0.5, 0.7, 0.1],
    [2.2, 2.9, 0.4],
    [1.9, 2.2, 0.3]
])

# Step 1: Zero-center the data
X_centered = X - np.mean(X, axis=0)

# Step 2: Compute Covariance Matrix
cov_matrix = np.cov(X_centered, rowvar=False)

# Step 3: Eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Step 4: Sort eigenvalues and eigenvectors in descending order
sort_idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[sort_idx]
eigenvectors = eigenvectors[:, sort_idx]

# Step 5: Project 3D data down to 2 Principal Components
top_2_components = eigenvectors[:, :2]
X_projected = X_centered @ top_2_components

print("=== PCA Transformation from Scratch ===")
print("Original 3D Shape :", X.shape)
print("Projected 2D Shape:", X_projected.shape)
print("\nExplained Variance Ratios:", (eigenvalues / np.sum(eigenvalues)).round(3))
print("First Component captures:", round((eigenvalues[0]/np.sum(eigenvalues))*100, 1), "% of total dataset variance!")
```

### 🔍 Line-by-Line Code Breakdown:
- 'X - np.mean(X, axis=0)': Essential! PCA assumes features are centered at the origin (0, 0).
- 'np.linalg.eig(cov_matrix)': Decomposes the covariance matrix into orthogonal vectors along the axes of maximum variation.
- 'X_centered @ top_2_components': Matrix multiplication projects the 3D data points onto the new 2D orthogonal subspace without losing the most critical information!
