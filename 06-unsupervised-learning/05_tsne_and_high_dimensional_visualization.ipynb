## 💡 1. PCA vs t-SNE: The Big Difference
- **PCA (Linear):** Focuses on preserving global variance. If your data is a Swiss Roll or spiral curve, PCA crushes the layers together into a mess.
- **t-SNE (Non-Linear):** Focuses on preserving **local neighborhoods** (points that are close together in 100D space stay close together in 2D plot!).
- *Golden Rule:* **t-SNE is strictly for human visual exploration in 2D/3D**. Do NOT use t-SNE output as input features to a classifier because it does not maintain global distances or orthogonal components!

```python
import numpy as np

# Probability definition in t-SNE: Convert Euclidean distances to Gaussian probabilities
def pairwise_affinities(X, sigma=1.0):
    # Squared distances
    sum_X = np.sum(np.square(X), 1)
    D = np.add(np.add(-2 * np.dot(X, X.T), sum_X).T, sum_X)
    # Gaussian kernel
    P = np.exp(-D / (2 * sigma ** 2))
    np.fill_diagonal(P, 0)
    P = P / np.sum(P)
    return P

pts = np.array([[0, 0], [0, 1], [10, 10], [10, 11]], dtype=np.float64)
affinity_matrix = pairwise_affinities(pts)
print("t-SNE High-Dimensional Probability Matrix P (Pairwise affinities):")
print(affinity_matrix.round(4))
print("\nNotice: Strong local probabilities between pts 0-1 and pts 2-3, near-zero cross-cluster probability!")
```
