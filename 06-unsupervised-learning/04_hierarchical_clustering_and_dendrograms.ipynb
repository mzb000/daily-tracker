## 💡 1. The Big Advantage over K-Means
In K-Means, you **MUST** specify $K$ in advance. If you guess wrong, the model splits or merges real groups incorrectly.
In **Hierarchical Clustering**, you don't choose $K$ upfront! The algorithm builds a complete **taxonomic family tree (Dendrogram)** of all points, and you can cut the tree at any horizontal height you desire!

```python
import numpy as np

# 5 points on a 1D line to easily visualize bottom-up merging
points = np.array([1, 2, 7, 8, 15])

# Pairwise distance matrix
dist_matrix = np.abs(points[:, np.newaxis] - points[np.newaxis, :])
print("Initial Pairwise Distance Matrix:")
print(dist_matrix)

print("\nStep 1: Point 1 and 2 are closest (Distance = 1). Merge them into Cluster (1,2)!")
print("Step 2: Point 7 and 8 are next closest (Distance = 1). Merge into Cluster (7,8)!")
print("Step 3: Point 15 merges last because it is an isolated outlier.")
```
