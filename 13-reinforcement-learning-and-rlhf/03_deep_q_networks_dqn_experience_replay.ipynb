```python
from collections import deque
import random
import numpy as np

# Experience Replay Memory Buffer from Scratch
class ReplayBuffer:
    def __init__(self, capacity=1000):
        self.buffer = deque(maxlen=capacity)

    def push(self, state, action, reward, next_state, done):
        self.buffer.append((state, action, reward, next_state, done))

    def sample(self, batch_size=4):
        return random.sample(self.buffer, batch_size)

    def __len__(self):
        return len(self.buffer)

memory = ReplayBuffer(capacity=100)
# Populate memory with transition tuples: (s, a, r, s', done)
for step in range(10):
    memory.push(state=step, action=step % 2, reward=1.0, next_state=step + 1, done=False)

sample_batch = memory.sample(batch_size=3)
print(f"Total Stored Transitions: {len(memory)}")
print("Random Mini-Batch Sampled from Replay Buffer (Breaks correlation!):")
for transition in sample_batch:
    print(" ", transition)
```
