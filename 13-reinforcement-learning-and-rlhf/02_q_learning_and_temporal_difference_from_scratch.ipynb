3. **The Q-Learning Update Rule:**
   $$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$
4. **The Exploration-Exploitation Dilemma ($\epsilon$-Greedy Strategy)**
5. **Complete Q-Learning Agent Implementation in Pure Python**

```python
import numpy as np

# 4 States, 2 Actions (0=Left, 1=Right)
n_states, n_actions = 4, 2
Q_table = np.zeros((n_states, n_actions))

alpha = 0.1   # Learning rate
gamma = 0.95  # Discount factor
epsilon = 0.2 # Exploration probability

# Simulate 100 training steps of Q-learning
np.random.seed(42)
for episode in range(100):
    state = 0
    while state < 3:
        # Epsilon-greedy action selection
        if np.random.rand() < epsilon:
            action = np.random.choice(n_actions)
        else:
            action = np.argmax(Q_table[state])

        # Environment dynamics: action 1 moves right, action 0 moves left
        next_state = min(3, state + 1) if action == 1 else max(0, state - 1)
        reward = 10.0 if next_state == 3 else 0.0

        # Q-Learning TD Update
        best_next_q = np.max(Q_table[next_state])
        td_target = reward + gamma * best_next_q
        td_error = td_target - Q_table[state, action]
        Q_table[state, action] += alpha * td_error

        state = next_state

print("=== Trained Q-Table (Rows=States 0-3, Cols=Left, Right) ===")
print(Q_table.round(2))
print("\nOptimal Policy (Argmax of Q-table for each state):")
for s in range(3):
    action_str = "MOVE RIGHT ->" if np.argmax(Q_table[s]) == 1 else "MOVE LEFT <-"
    print(f"State {s}: Best Action is [{action_str}]")
```
