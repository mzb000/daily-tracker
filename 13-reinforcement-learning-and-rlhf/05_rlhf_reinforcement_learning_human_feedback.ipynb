4. **Hands-On Reward Model Loss Implementation**

```python
import numpy as np

# Bradley-Terry Preference Model from Scratch
# Human preference: Response A is preferred over Response B (A > B)
def bradley_terry_loss(reward_chosen, reward_rejected):
    # Loss = -log(sigmoid(r_chosen - r_rejected))
    diff = reward_chosen - reward_rejected
    prob = 1.0 / (1.0 + np.exp(-diff))
    loss = -np.log(prob + 1e-10)
    return loss, prob

# Case 1: Model assigns higher reward to chosen answer (+2.5 vs -1.0)
loss_good, prob_good = bradley_terry_loss(reward_chosen=2.5, reward_rejected=-1.0)

# Case 2: Model assigns wrong reward (prefers rejected toxic answer!)
loss_bad, prob_bad = bradley_terry_loss(reward_chosen=-1.5, reward_rejected=2.0)

print("=== Reward Model Optimization ===")
print(f"Correct Reward Ranking -> Loss: {loss_good:.4f} | Confidence Chosen is Better: {prob_good*100:.1f}%")
print(f"Wrong Reward Ranking   -> Loss: {loss_bad:.4f}  | High Penalty to force alignment!")
```
