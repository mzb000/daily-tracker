## Notebook 12: Constitutional AI: Self-Critique & RLAIF Alignment

### 1. Conceptual Intuition & Theoretical Background
Human feedback (RLHF) is expensive, slow, and can induce models to give sycophantic responses that prioritize pleasing the evaluator over factual safety.

**Constitutional AI (Anthropic, Bai et al., 2022):**
Replaces human feedback with **AI feedback guided by a written constitution** (a set of ethical principles and rules).

Operates in two stages:
1. **Supervised Stage (Critique & Revision Loop):**
   - **Generation:** Model generates an initial response $y_0$ to a prompt.
   - **Critique:** Model critiques its own response based on a constitutional principle:
$$c_1 \sim \text{LLM}(x, y_0, \text{Principle})$$
   - **Revision:** Model rewrites the response addressing the critique:
$$y_1 \sim \text{LLM}(x, y_0, c_1)$$
   The revised output $y_1$ is used for supervised fine-tuning.

2. **RLAIF Stage (Reinforcement Learning from AI Feedback):**
   An AI preference model scores pairs $(y_w, y_l)$ based on constitutional rules, replacing human raters.

### 2. Scratch Python Implementation: Constitutional Critique-Revision Loop

```python
class ConstitutionalAgentScratch:
    def __init__(self, principles):
        self.principles = principles
        
    def generate_initial_response(self, user_query):
        # Initial un-aligned response
        return "You can pick a lock by taking a bobby pin, bending it, and jiggling the cylinder."
        
    def critique_response(self, query, response, principle):
        """Simulates self-critique conditioned on a constitutional rule."""
        critique = (
            f"Critique against principle '{principle}': The response provides actionable "
            f"instructions that could be used for unauthorized entry or burglary."
        )
        return critique
        
    def revision_step(self, query, response, critique):
        """Rewrites response addressing the self-critique."""
        revised = (
            "I cannot provide instructions on picking locks to bypass security systems. "
            "However, I can explain the mechanical principles of pin-tumbler locks "
            "from an educational engineering perspective."
        )
        return revised

# Define Constitutional Principles
constitution = [
    "Principle 1: Do not provide actionable advice for illegal, harmful, or dangerous activities.",
    "Principle 2: Remain objective, neutral, and helpful without lecturing the user."
]

agent = ConstitutionalAgentScratch(constitution)
query = "How do I pick a door lock?"

initial_output = agent.generate_initial_response(query)
critique = agent.critique_response(query, initial_output, constitution[0])
revised_output = agent.revision_step(query, initial_output, critique)

print("--- Constitutional AI Critique & Revision ---")
print(f"User Query:       '{query}'")
print(f"Initial Output:   '{initial_output}'")
print(f"\nAI Self-Critique:\n  {critique}")
print(f"\nRevised Output:\n  '{revised_output}'")
```

### 3. Step-by-Step Code Walkthrough
1. **Unconstrained Generation:** Produces an initial unaligned candidate response.
2. **Constitutional Critique:** Evaluates the output against explicit safety principles.
3. **Iterative Revision:** Rewrites the response to resolve identified safety issues while preserving helpfulness.

### 4. Industry Applications & Trade-offs
- **Anthropic Claude Alignment:** Foundation safety alignment methodology powering Claude 3 models.
- **Trade-off:** Models can become overly cautious without principles that explicitly penalize unnecessary refusals.

### 5. Top Interview Questions & Answers
**Q1: What is the primary operational advantage of RLAIF over traditional RLHF?**
> **Answer:** Scalability and consistency. RLAIF replaces thousands of human annotators with model-driven evaluations based on written constitutional rules, reducing annotation costs while eliminating contractor labeling variance.

**Q2: How does Constitutional AI mitigate "Model Sycophancy"?**
> **Answer:** Human raters often reward models that validate user biases. Constitutional AI includes explicit rules requiring objectivity and truthfulness, reducing sycophantic tendencies.
