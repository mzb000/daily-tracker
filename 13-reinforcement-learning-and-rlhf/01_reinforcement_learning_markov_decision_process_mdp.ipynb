   $$V(s) = \max_a \left[ R(s, a) + \gamma \sum_{s'} P(s' | s, a) V(s') \right]$$
5. **Hands-On Gridworld State-Value Calculation in NumPy**

```python
import numpy as np

# 3-State Line World: State 0 <-> State 1 <-> State 2 (Goal: +10 reward)
# Let's compute State Values using Bellman Value Iteration
states = [0, 1, 2]
V = np.zeros(3)  # Initial values: [0, 0, 0]
gamma = 0.9      # Discount factor

# Value Iteration for 10 rounds
for iteration in range(10):
    V_new = np.copy(V)
    # State 2 is terminal goal: Reward = 10
    V_new[2] = 10.0
    # State 1 can transition to State 2 (Reward 0 + gamma * V[2])
    V_new[1] = 0.0 + gamma * V[2]
    # State 0 can transition to State 1 (Reward 0 + gamma * V[1])
    V_new[0] = 0.0 + gamma * V[1]
    V = V_new

print("=== Converged Bellman State Values ===")
print("State 0 Value :", round(V[0], 2), "(2 steps away from reward)")
print("State 1 Value :", round(V[1], 2), "(1 step away from reward)")
print("State 2 Value :", round(V[2], 2), "(The goal state!)")
```
