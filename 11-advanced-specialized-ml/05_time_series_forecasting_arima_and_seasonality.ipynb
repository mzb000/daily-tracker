```python
import numpy as np

# Non-stationary trending series: stock price drifting upwards
time_index = np.arange(10)
non_stationary_stock = 100.0 + 3.5 * time_index + np.random.normal(0, 1, 10)

# First-Order Differencing: d = 1 -> y_t - y_{t-1}
stationary_diff = np.diff(non_stationary_stock)

print("Original Trending Stock Price:\n", non_stationary_stock.round(1))
print("\nFirst Differenced Series (Trend removed, stationary around mean 3.5!):\n", stationary_diff.round(2))
```
