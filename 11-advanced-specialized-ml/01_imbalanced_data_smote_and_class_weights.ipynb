```python
import numpy as np

# SMOTE Interpolation from Scratch for a 2D minority sample
def smote_synthetic_sample(minority_point, neighbor_point):
    diff = neighbor_point - minority_point
    gap = np.random.rand()
    return minority_point + gap * diff

p1 = np.array([2.0, 3.0])  # Real minority sample
p2 = np.array([2.5, 3.8])  # Its nearest neighbor in minority class

np.random.seed(42)
synth_sample = smote_synthetic_sample(p1, p2)
print("Real Sample 1      :", p1)
print("Real Sample 2 (KNN):", p2)
print("Synthetic Generated Sample (along line segment):", synth_sample.round(3))
```
