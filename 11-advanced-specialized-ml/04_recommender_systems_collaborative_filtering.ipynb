```python
import numpy as np
import pandas as pd

# User-Movie Ratings Matrix (1 to 5 stars, 0 = unrated)
ratings = pd.DataFrame({
    "Interstellar": [5, 4, 1, 0],
    "Inception":    [5, 5, 0, 1],
    "The Godfather":[0, 1, 5, 4],
    "Goodfellas":   [1, 0, 5, 5]
}, index=["User_A", "User_B", "User_C", "User_D"])

print("User-Item Ratings Matrix:")
print(ratings)

# Item-Based Cosine Similarity between Interstellar and Inception
v_interstellar = ratings["Interstellar"].values
v_inception = ratings["Inception"].values
sim_scifi = np.dot(v_interstellar, v_inception) / (np.linalg.norm(v_interstellar) * np.linalg.norm(v_inception))

v_godfather = ratings["The Godfather"].values
sim_cross_genre = np.dot(v_interstellar, v_godfather) / (np.linalg.norm(v_interstellar) * np.linalg.norm(v_godfather))

print(f"\nSimilarity(Interstellar, Inception)   : {sim_scifi:.3f} (High affinity!)")
print(f"Similarity(Interstellar, The Godfather): {sim_cross_genre:.3f} (Different movie taste)")
```
