```python
import numpy as np

# Simulation: Outliers isolate in very few random cuts!
data_points = np.array([50, 52, 49, 51, 53, 50, 500]) # 500 is extreme outlier

# Random cut between min and max
np.random.seed(42)
cut = np.random.uniform(np.min(data_points), np.max(data_points))
print(f"Random Cut Value: {cut:.1f}")
print("Left Group  (Normal points clustered):", data_points[data_points < cut])
print("Right Group (Outlier isolated on first cut!):", data_points[data_points >= cut])
```
