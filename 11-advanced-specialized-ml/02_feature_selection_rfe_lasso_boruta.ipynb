```python
import numpy as np

# The Boruta "Shadow Feature" Concept from Scratch
# Create shadow features by randomly shuffling real features
X_real = np.array([
    [10.0, 200.0],
    [12.0, 150.0],
    [15.0, 300.0],
    [18.0, 250.0]
])

np.random.seed(42)
X_shadow = np.copy(X_real)
for col in range(X_shadow.shape[1]):
    np.random.shuffle(X_shadow[:, col])  # Break correlation with target!

X_boruta_matrix = np.hstack([X_real, X_shadow])
print("Original Features Matrix Shape:", X_real.shape)
print("Boruta Matrix Shape (Real + Shadow):", X_boruta_matrix.shape)
print("Concept: If a real feature cannot beat the maximum importance of random shadow features, discard it!")
```
