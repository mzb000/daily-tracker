## 🎯 The Goal of Linear Regression
Fit a line through data points such that the total prediction error is minimized.
$$\hat{y} = w \cdot x + b$$

### Loss Function: Mean Squared Error (MSE)
$$J(w, b) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2$$

### Gradients (Derivatives):
$$\frac{\partial J}{\partial w} = \frac{1}{m} \sum (\hat{y}^{(i)} - y^{(i)}) x^{(i)}$$
$$\frac{\partial J}{\partial b} = \frac{1}{m} \sum (\hat{y}^{(i)} - y^{(i)})$$

```python
import numpy as np

class LinearRegressionScratch:
    def __init__(self, lr=0.01, epochs=1000):
        self.lr = lr
        self.epochs = epochs
        self.w = 0.0
        self.b = 0.0
        self.history = []

    def fit(self, X, y):
        n = len(X)
        for _ in range(self.epochs):
            # Forward pass
            y_pred = self.w * X + self.b
            loss = np.mean((y_pred - y) ** 2)
            self.history.append(loss)

            # Gradient step
            dw = (2 / n) * np.sum((y_pred - y) * X)
            db = (2 / n) * np.sum(y_pred - y)

            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict(self, X):
        return self.w * X + self.b

# Synthetic test data: True relation is y = 2.5 * x + 5.0
np.random.seed(42)
X = np.linspace(1, 10, 50)
y = 2.5 * X + 5.0 + np.random.normal(0, 0.5, 50)

model = LinearRegressionScratch(lr=0.01, epochs=2000)
model.fit(X, y)

print(f"Trained Slope (w) : {model.w:.3f} (True ~ 2.5)")
print(f"Trained Bias  (b) : {model.b:.3f} (True ~ 5.0)")
```
