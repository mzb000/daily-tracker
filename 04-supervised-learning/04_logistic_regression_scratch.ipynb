## 🎯 From Numbers to Probabilities
Linear regression predicts continuous values $(-\infty, +\infty)$.
Logistic regression passes the linear output through the **Sigmoid function**:
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$
Squashing any real number into a valid probability between $0.0$ and $1.0$!

```python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))

class LogisticRegressionScratch:
    def __init__(self, lr=0.1, epochs=1000):
        self.lr = lr
        self.epochs = epochs
        self.w = None
        self.b = 0.0

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features)
        self.b = 0.0

        for _ in range(self.epochs):
            # Compute probabilities
            z = np.dot(X, self.w) + self.b
            y_pred = sigmoid(z)

            # Gradient update
            dw = (1 / n_samples) * np.dot(X.T, (y_pred - y))
            db = (1 / n_samples) * np.sum(y_pred - y)

            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict(self, X):
        probs = sigmoid(np.dot(X, self.w) + self.b)
        return (probs >= 0.5).astype(int)

# 4 simple 2D training samples: Class 0 vs Class 1
X_train = np.array([[1.0, 1.0], [2.0, 1.5], [8.0, 9.0], [9.0, 8.5]])
y_train = np.array([0, 0, 1, 1])

clf = LogisticRegressionScratch(lr=0.5, epochs=2000)
clf.fit(X_train, y_train)

predictions = clf.predict(X_train)
print("True Labels       :", y_train)
print("Model Predictions :", predictions)
print("Accuracy          :", np.mean(predictions == y_train) * 100, "%")
```
