## 🎯 The Overfitting Problem
When models have many collinear features, weights explode to astronomical values (e.g., $+10,000$ on one feature, $-9,999$ on another), making predictions volatile.
Regularization adds a cost penalty on large weights:
- **Ridge ($L_2$):** Penalizes $\lambda \sum w_i^2$. Shrinks weights smoothly toward zero.
- **Lasso ($L_1$):** Penalizes $\lambda \sum |w_i|$. Forces unimportant weights to **exactly zero** (Sparse Solutions).

```python
import numpy as np

def compute_regularized_loss(y_true, y_pred, weights, alpha=0.1):
    mse = np.mean((y_true - y_pred) ** 2)
    l1 = alpha * np.sum(np.abs(weights))
    l2 = alpha * np.sum(weights ** 2)

    return {
        "Base MSE Loss": round(mse, 4),
        "Ridge Total Loss (MSE + L2)": round(mse + l2, 4),
        "Lasso Total Loss (MSE + L1)": round(mse + l1, 4)
    }

y_true = np.array([10.0, 20.0, 30.0])
y_pred = np.array([9.8, 20.4, 29.7])
weights = np.array([3.2, -0.05, 4.1, -1.8])

losses = compute_regularized_loss(y_true, y_pred, weights, alpha=0.5)
for k, v in losses.items():
    print(f"{k:28}: {v}")
```
