## 🎯 Which Metric Should You Use?
- **MAE:** Best when you want an intuitive average error in real-world units, robust to occasional outliers.
- **RMSE:** Best when large mistakes are disastrous (e.g., flight arrival delay, patient drug dosage).
- **$R^2$:** Normalized score between 0 and 1 (1.0 = perfect model, 0.0 = model performs no better than guessing the mean).

```python
import numpy as np

def evaluate_regression(y_true, y_pred):
    mae = np.mean(np.abs(y_true - y_pred))
    mse = np.mean((y_true - y_pred) ** 2)
    rmse = np.sqrt(mse)
    
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    ss_res = np.sum((y_true - y_pred) ** 2)
    r2 = 1.0 - (ss_res / ss_tot)
    
    return {"MAE": mae, "MSE": mse, "RMSE": rmse, "R2_Score": r2}

y_actual = np.array([100.0, 150.0, 200.0, 250.0, 300.0])
y_predicted = np.array([105.0, 148.0, 195.0, 255.0, 310.0])

metrics = evaluate_regression(y_actual, y_predicted)
for m, val in metrics.items():
    print(f"{m:10}: {val:.4f}")
```
