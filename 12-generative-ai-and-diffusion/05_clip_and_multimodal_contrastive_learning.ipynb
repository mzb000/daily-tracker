```python
import numpy as np

# Dual-Encoder Embedding Space: 3 Images and 3 Text Descriptions
# Image Embeddings (Shape: 3 x 4, L2 Normalized)
img_embeddings = np.array([
    [0.9, 0.1, 0.2, 0.0],  # Image 0: Photo of a Golden Retriever dog
    [0.1, 0.8, 0.1, 0.3],  # Image 1: Photo of a sports car
    [0.0, 0.2, 0.9, 0.1]   # Image 2: Photo of an airplane in sky
])
img_embeddings /= np.linalg.norm(img_embeddings, axis=1, keepdims=True)

# Text Embeddings from text encoder (Shape: 3 x 4, L2 Normalized)
text_prompts = [
    "a photo of a playful dog",
    "a photo of a fast red sports car",
    "a commercial airplane flying in clouds"
]
text_embeddings = np.array([
    [0.88, 0.12, 0.18, 0.05],  # Matches Dog
    [0.15, 0.85, 0.08, 0.25],  # Matches Car
    [0.02, 0.18, 0.89, 0.12]   # Matches Airplane
])
text_embeddings /= np.linalg.norm(text_embeddings, axis=1, keepdims=True)

# Cross-Modal Dot Product Cosine Similarity Matrix: Shape (3, 3)
logits = img_embeddings @ text_embeddings.T

# Softmax over text prompts for each image
exp_logits = np.exp(logits * 10.0)  # Temperature scaling tau=10
zero_shot_probs = exp_logits / np.sum(exp_logits, axis=1, keepdims=True)

print("=== CLIP Zero-Shot Classification Probabilities ===")
for img_id in range(3):
    best_text_id = np.argmax(zero_shot_probs[img_id])
    confidence = zero_shot_probs[img_id, best_text_id] * 100
    print(f"Image {img_id} -> Predicted: '{text_prompts[best_text_id]}' ({confidence:.1f}% Confidence)")
```
