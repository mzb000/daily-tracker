   $$x_t = \sqrt{\bar{\alpha}_t} x_0 + \sqrt{1 - \bar{\alpha}_t} \epsilon, \quad \epsilon \sim \mathcal{N}(0, I)$$

```python
import numpy as np

# Forward Diffusion Process: Linear Variance Schedule (Beta: 0.0001 to 0.02 over T=1000 steps)
T = 1000
beta = np.linspace(1e-4, 0.02, T)
alpha = 1.0 - beta
alpha_bar = np.cumprod(alpha)  # Cumulative product of alpha

# Closed-form jump to timestep t=500 without looping!
def add_noise_to_image(x_0, t_step):
    eps = np.random.randn(*x_0.shape)
    sqrt_alpha_bar = np.sqrt(alpha_bar[t_step])
    sqrt_one_minus_alpha_bar = np.sqrt(1.0 - alpha_bar[t_step])
    # x_t = sqrt(alpha_bar_t) * x_0 + sqrt(1 - alpha_bar_t) * eps
    x_t = sqrt_alpha_bar * x_0 + sqrt_one_minus_alpha_bar * eps
    return x_t, eps

# Synthetic 1D pixel value: 100.0
clean_pixel = np.array([100.0])

noisy_t100, _ = add_noise_to_image(clean_pixel, 100)
noisy_t500, _ = add_noise_to_image(clean_pixel, 500)
noisy_t999, _ = add_noise_to_image(clean_pixel, 999)

print("Original Clean Signal :", clean_pixel[0])
print("Signal at Timestep 100:", round(noisy_t100[0], 2), "(Slight noise)")
print("Signal at Timestep 500:", round(noisy_t500[0], 2), "(Heavy noise)")
print("Signal at Timestep 999:", round(noisy_t999[0], 2), "(Pure Gaussian static noise!)")
```
