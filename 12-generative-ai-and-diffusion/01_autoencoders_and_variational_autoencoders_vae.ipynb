4. **The Reparameterization Trick: Making Random Sampling Differentiable!**
   $$z = \mu + \sigma \odot \epsilon, \quad \epsilon \sim \mathcal{N}(0, I)$$
5. **The VAE Loss: Reconstruction Loss + KL-Divergence**
6. **Hands-On NumPy Implementation & Latent Space Interpolation**

```python
import numpy as np

# Demonstrating the Reparameterization Trick from scratch
# Without this trick, backpropagation cannot pass through a random node!
def reparameterize(mu, log_var):
    std = np.exp(0.5 * log_var)
    eps = np.random.randn(*mu.shape)  # Random noise isolated from gradients
    return mu + std * eps

# Simulated encoder output for 2 samples in 3D latent space
mu_encoder = np.array([[0.5, -0.2, 1.2], [-0.8, 0.4, -0.1]])
log_var_encoder = np.array([[0.0, -0.5, 0.2], [-0.1, 0.1, 0.0]])

# Sample latent vectors z
z_latent = reparameterize(mu_encoder, log_var_encoder)

# Compute KL Divergence Loss analytically: D_KL = -0.5 * sum(1 + log(sigma^2) - mu^2 - sigma^2)
kl_loss = -0.5 * np.sum(1 + log_var_encoder - np.square(mu_encoder) - np.exp(log_var_encoder), axis=1)

print("Encoder Predicted Means (mu):\n", mu_encoder)
print("\nSampled Latent Vectors (z):\n", z_latent.round(3))
print("\nKL Divergence Penalty per sample:", kl_loss.round(3))
print("Notice: KL loss penalizes deviations from the standard normal N(0, I) prior!")
```
