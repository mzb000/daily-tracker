```python
import numpy as np

# Simulating a small image: 32x32 pixels, 3 channels (RGB)
H, W, C = 32, 32, 3
patch_size = 8  # 8x8 patches
d_model = 16    # Embedding dimension

# Number of patches: (32/8) * (32/8) = 4 * 4 = 16 patches
num_patches = (H // patch_size) * (W // patch_size)
patch_dim = patch_size * patch_size * C  # 8 * 8 * 3 = 192 numbers per patch

# Simulated random image
image = np.random.randn(H, W, C)

# Extract 16 patches by reshaping
patches = []
for i in range(0, H, patch_size):
    for j in range(0, W, patch_size):
        patch = image[i : i + patch_size, j : j + patch_size, :]
        patches.append(patch.flatten())

patches = np.array(patches)  # Shape: (16, 192)

# Linear Projection Layer: W_e of shape (192, 16)
W_proj = np.random.randn(patch_dim, d_model) * 0.02
patch_tokens = patches @ W_proj  # Shape: (16, 16)

# Prepend [CLS] token of shape (1, 16)
cls_token = np.zeros((1, d_model))
all_tokens = np.vstack([cls_token, patch_tokens])  # Shape: (17, 16)

print("Original Image Shape       :", image.shape)
print("Extracted Patches Shape    :", patches.shape, f"({num_patches} patches of {patch_dim} pixels)")
print("Linear Projected Tokens    :", patch_tokens.shape)
print("Final Input with [CLS] Token:", all_tokens.shape, "(Ready for Standard Transformer Self-Attention!)")
```
