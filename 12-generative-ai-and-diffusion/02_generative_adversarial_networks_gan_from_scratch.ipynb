```python
import numpy as np

# 1D GAN Intuition: Generator learns to transform Gaussian noise into real data distribution (Mean=4.0, Std=0.5)
def real_data_sampler(batch_size=32):
    return np.random.normal(loc=4.0, scale=0.5, size=(batch_size, 1))

def noise_sampler(batch_size=32):
    return np.random.uniform(low=-1.0, high=1.0, size=(batch_size, 1))

# Simple Generator parameter: single weight and bias: G(z) = w_g * z + b_g
w_g = 1.0
b_g = 0.0

z = noise_sampler(5)
fake_samples = w_g * z + b_g
real_samples = real_data_sampler(5)

print("Real Data Distribution Sample (Target Mean ~4.0):\n", real_samples.round(2).flatten())
print("\nInitial Fake Samples from Generator (Before Training):\n", fake_samples.round(2).flatten())
print("\nGenerator Objective: Shift w_g and b_g until fake distribution matches real distribution!")
```
