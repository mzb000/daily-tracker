## 🎯 Why is EDA the Most Important Step in ML?
"Garbage In, Garbage Out." A fancy Neural Network trained on dirty, skewed data will make garbage predictions.
EDA (Exploratory Data Analysis) allows us to:
1. Understand distribution shapes (Normal, Right-Skewed, Bi-Modal).
2. Measure central tendencies (Mean vs Median).
3. Detect data leakage and measurement errors before training!

```python
import numpy as np
import pandas as pd

np.random.seed(42)
# Simulating e-commerce order values (heavily right-skewed)
order_values = np.random.exponential(scale=50.0, size=1000)

df = pd.DataFrame({"order_value": order_values})
print("Mean Order Value   : $", round(df["order_value"].mean(), 2))
print("Median Order Value : $", round(df["order_value"].median(), 2))
print("Skewness (>0 right):", round(df["order_value"].skew(), 2))
```
